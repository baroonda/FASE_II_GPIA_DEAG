```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
import pickle

# 1. Cargar el dataset real de Kaggle
print("Cargando dataset...")
df = pd.read_csv('/content/creditcard.csv')

# 2. Separar características (Time, V1 a V28, Amount) y la etiqueta (Class)
X = df.drop('Class', axis=1)
y = df['Class']

# 3. Entrenar el modelo
# (Limitamos los estimadores y profundidad para que el entrenamiento sea rápido
# y el archivo .pkl no quede demasiado pesado para subirlo a la terminal después)
print("Entrenando Random Forest...")
modelo = RandomForestClassifier(n_estimators=50, max_depth=5, random_state=42)
modelo.fit(X, y)

# 4. Exportar el modelo real a un archivo físico
with open('model.pkl', 'wb') as f:
    pickle.dump(modelo, f)

print("¡Listo! Modelo real entrenado y guardado.")

# 5. Descargar el archivo a tu computadora
from google.colab import files
files.download('model.pkl')
```
